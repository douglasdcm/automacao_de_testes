# Classes e objetos

# Decorator

```python
def deco(func, *args):
  print(f"antes da func '{func.__name__}'")
  func(*args)
  print(f"depois da func '{func.__name__}'")

def func1(x,y):
  print(f"dentro da func1'")

func1(2,3)
print("-----")
deco(func1, 2, 3)
```

```python
def deco(func, *args):
  def inner(*args):
    print(f"antes da func '{func.__name__}'")
    r = func(*args)
    print(f"Resultado {r}")
    print(f"depois da func '{func.__name__}'")
    return r
  return inner

@deco
def func1(x,y):
  return x + y

func1(1, 3)
```

# Exceptions

```python
class CalculadoraFraca:
  def somar(self, *args):
    for n in args:
      n+=n
    return n

  def subtrair(self, *args):
    raise NotImplementedError("Nao sei fazer isso ainda")

  def multiplicar(self, *args):
    raise NotImplementedError("Nao sei fazer isso ainda")

  def dividir(self, *args):
    raise NotImplementedError("Nao sei fazer isso ainda")
```

```python
c = CalculadoraFraca()
resultado = c.somar(1,2)
print(resultado)
```

```python
# Exceptions
c.subtrair(1,2,3)
```

```python
# Try-except
try:
  c.subtrair(6,7,8)
except NotImplementedError as e:
  print(str(e))
```
